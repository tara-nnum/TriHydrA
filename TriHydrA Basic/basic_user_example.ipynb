# TriHydrA Basic example
Run each cell from top to bottom. The final cell creates a `results` folder containing TXT reports and interactive HTML plots.

```python
from pathlib import Path
import pandas as pd
from trihydra import run_trihydra_network, save_results, plot_results
```

```python
HERE = Path.cwd()
DATA_FILE = HERE / 'example_three_stations.csv'
CONTEXT_FILE = HERE / 'context.csv'
RESULTS_DIRECTORY = HERE / 'results'
UNIT = 'mm/day'
```

```python
data = pd.read_csv(DATA_FILE, parse_dates=['date']).set_index('date')
data.head()
```

```python
stations = {station_id: data[station_id] for station_id in data.columns}
result = run_trihydra_network(
    stations,
    context_path=CONTEXT_FILE,
    target_station_ids=list(stations),
    unit=UNIT,
)
result.summary
```

```python
save_results(result, RESULTS_DIRECTORY)
plot_results(result, RESULTS_DIRECTORY)
print(f'Finished. Open: {RESULTS_DIRECTORY}')
```
